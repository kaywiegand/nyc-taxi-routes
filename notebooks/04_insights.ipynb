# Business Communication & Insights

## Executive Summary

Von 293.369 bereinigten Fahrten (2016er NYC-Taxidaten) starten nur **1,93 %** am Flughafen JFK
(5.649 Fahrten). Zählt man Fahrten, die JFK entweder als Start- oder Zielpunkt haben, sind es
**2,6 % (7.613 Fahrten)** — diese erzeugen aber **9,99 % des gesamten Fahrpreis-Umsatzes**: eine
JFK-Fahrt ist im Schnitt fast **4× so wertvoll** wie eine durchschnittliche Fahrt.

Die JFK-Nachfrage ist nicht gleichmäßig verteilt: **Montags** ist der JFK-Anteil mit **2,60 %**
am höchsten, **samstags** mit **1,44 %** am niedrigsten. Über den Tag verteilt liegt die
Gesamt-Peak-Stunde bei **18 Uhr**, die JFK-Peak-Stunde bereits bei **17 Uhr** — und in den frühen
Morgenstunden (5–6 Uhr) ist der JFK-Anteil **doppelt so hoch** wie der allgemeine Verkehrsanteil zu
dieser Zeit (vermutlich frühe Flugabflüge).

**Empfehlung:** keine statische Flottenzuteilung für JFK, sondern eine Zeit- und
Wochentag-gewichtete Präsenz — siehe unten.

## Next Steps & Open Questions

* Individualisierung der Visualisierungen (Aufgabe 8) — Styling/Branding auf Auftraggeber zuschneiden
* Absolute Flottengröße ist im Datensatz nicht enthalten — die obigen Faktoren sind relative
  Gewichtungen, keine konkreten Taxi-Stückzahlen. Für eine harte Stückzahl-Empfehlung würde die
  Gesamtflottengröße des Unternehmers benötigt.
* Saisonale Effekte (Feiertage, Events am Flughafen) sind im 2016er Datensatz nicht auflösbar —
  offene Frage für Folgeanalysen
